```python
# Load env variables and create API client
from dotenv import load_dotenv
load_dotenv()


from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
# Helper functions
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages,system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    if system:
        params["system"] = system

    message = client.messages.create(**params)
    for block in message.content:
        if block.type == "text":
            return block.text
```

```python
messages = []

system = """
    Sos un experto desarrollador de Python.
    Quiero que des respuestas cortas y concisas.
    Sin comentarios o caracteres especiales.
"""
add_user_message(messages, "Escribi una funcion en Python que chequee si un string tiene duplicado caracteres.")

answer = chat(messages,system)

answer
```
