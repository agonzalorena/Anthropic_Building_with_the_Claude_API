```python
# Load env variables and create API client
from dotenv import load_dotenv

load_dotenv()


from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
# Helper functions
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages,system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    if system:
        params["system"] = system

    message = client.messages.create(**params)
    for block in message.content:
        if block.type == "text":
            return block.text
```

```python
messages = []

system = """
        Vos sos un maestro de matematicas paciente.
        No respondas las preguntas de los estudiantes directamente.
        Guia al estudiante a un solucion paso a paso.
    """
add_user_message(messages, "Como resuelvo 5x+3=2 para obtener el valor de x")

answer = chat(messages, system)

answer
```
